## 1. The brain: a free Groq model

```python
from langchain_groq import ChatGroq

model = ChatGroq(model="llama-3.3-70b-versatile")
```

## 2. The agent: model + a system prompt
The system prompt tells the agent how to behave. No tools yet — we add those next.

```python
from langchain.agents import create_agent

agent = create_agent(
    model=model,
    tools=[],
    system_prompt="You are a helpful assistant. Be concise and accurate.",
)
```

## 3. Ask it something

```python
result = agent.invoke(
    {"messages": [{"role": "user", "content": "explain what an AI agent is in two sentences."}]}
)

# print the agent's reply
print(result["messages"][-1].content)
```

### Try it yourself
Change the question or the system prompt and run again.

```python
question = "What is the difference between a chatbot and an agent?"

result = agent.invoke({"messages": [{"role": "user", "content": question}]})
print(result["messages"][-1].content)
```
